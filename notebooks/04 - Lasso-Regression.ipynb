## 2. What Is Lasso Regression?

Lasso (Least Absolute Shrinkage and Selection Operator) is a regularized version of Multiple Linear Regression.

Like ordinary linear regression, it models the relationship between multiple independent variables (features) and one dependent variable (target).

However, it introduces an additional penalty based on the absolute values of the model coefficients.

The mathematical representation is:

\[
$\hat{y} = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b$
\]

The main difference is that Lasso penalizes large coefficients and can force some of them to become exactly zero.

**Why do we need Lasso Regression?**

Ordinary Linear Regression can suffer from:

- Overfitting: The model learns the training data too closely and performs poorly on unseen data.
- Irrelevant features: Some features may contribute little or no useful information to predictions.
- High model complexity: Using too many features can make the model unnecessarily complex.
- Multicollinearity: Highly correlated features can make coefficient estimates unstable.

Lasso addresses these problems by introducing L1 Regularization.

## 3. Mathematical Foundations

### 3.1. Mean Squared Error (MSE)

Ordinary Linear Regression minimizes the Mean Squared Error:

\[
$MSE(W,b) = \frac{1}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})^2$
\]

Where:

- \$(m\$): Number of training examples.
- \$(\hat{y}^{(i)}\$): Predicted value.
- \$(y^{(i)}\$): Actual target value.
- \$(W\$): Vector of model weights.
- \$(b\$): Bias (intercept).

### 3.2. L1 Regularization Penalty

Lasso introduces a penalty based on the sum of the absolute values of the model weights:

\[
$L1(W) = \sum_{j=1}^{n}|w_j|$
\]

Unlike L2 Regularization, which penalizes squared coefficients, L1 Regularization penalizes their absolute values.

This penalty encourages sparsity, meaning that some feature coefficients can become exactly zero.

### 3.3. Lasso Regression Cost Function

The Lasso objective combines prediction error and the L1 penalty:

\[
$J(W,b) = MSE(W,b) + \lambda\sum_{j=1}^{n}|w_j|$
\]

Where:

- \$(MSE\$): Measures prediction error.
- \$(\lambda\$): Regularization parameter controlling the strength of the penalty.
- \$(W\$): Vector of model weights.

**Important:**

- When \$(\lambda = 0\$), Lasso becomes ordinary linear regression.
- When \$(\lambda\$) increases, the model applies stronger regularization.
- Some coefficients can become exactly zero.
- The bias (intercept) is not penalized.
- Features whose coefficients become zero are effectively excluded from the model.

### 3.4. Lasso Regression and Gradient Descent

To optimize the Lasso cost function, we need to calculate its gradients.

First, calculate predictions:

\[
$\hat{y}=XW+b$
\]

The cost function is:

\[
$J(W,b)=\frac{1}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})^2+\lambda\sum_{j=1}^{n}|w_j|$
\]

The derivative of the absolute value function is not defined at zero.

For nonzero coefficients, its derivative is:

\[
$\frac{d|w_j|}{dw_j} =
\begin{cases}
1 & \text{if } w_j > 0\\
-1 & \text{if } w_j < 0
\end{cases}$
\]

At zero, we use a subgradient, which can take any value between -1 and 1.

For a basic subgradient implementation, we can choose zero at the origin:

\[
$\operatorname{sign}(w_j)=
\begin{cases}
1 & \text{if } w_j > 0\\
0 & \text{if } w_j = 0\\
-1 & \text{if } w_j < 0
\end{cases}$
\]

**Weight subgradient:**

\[
$\frac{\partial J}{\partial W} =$
$\frac{2}{m}X^T(\hat{y}-y)+\lambda\operatorname{sign}(W)$
\]

**Bias gradient:**

\[
$\frac{\partial J}{\partial b} =$
$\frac{2}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})$
\]

**Parameter updates:**

\[
$W := W-\alpha\left[
\frac{2}{m}X^T(\hat{y}-y)+\lambda\operatorname{sign}(W)
\right]$
\]

\[
$b := b-\alpha\frac{2}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})$
\]

Where \$(\alpha\$) represents the learning rate.

**Note:** Subgradient descent is a possible optimization approach, but it does not guarantee exact zero coefficients in finite iterations. Coordinate descent or proximal gradient descent with soft-thresholding is particularly useful for obtaining sparse Lasso solutions.

### 3.5. Soft-Thresholding Operator

The soft-thresholding operator is an important concept in Lasso optimization.

It helps shrink coefficients toward zero and sets sufficiently small coefficients exactly to zero.

\[
$S_{\tau}(z)=
\begin{cases}
z-\tau & \text{if } z>\tau\\
0 & \text{if } |z|\leq\tau\\
z+\tau & \text{if } z<-\tau
\end{cases}$
\]

Where:

- \$(z\$): A coefficient before thresholding.
- \$(\tau\$): Threshold determined by the regularization strength.

This operator explains why Lasso can perform automatic feature selection.

```python
import pandas as pd
import numpy as np
import math
from pathlib import Path

cwd = Path.cwd()
root = cwd.parent
data_dir = root / "Datasets" / "Lasso-regression"

df = pd.read_csv(data_dir / "lasso_regression_dataset.csv")

print(df.info())
print(df.describe())
print(df.shape)
```

```python
y = df["performance_score"]
X = df.drop(columns=["performance_score"])

def train_test_split(X,y,test_size=0.2):
    train_idx = X.sample(frac=1-test_size,random_state=42).index

    X_train = X.loc[train_idx]
    y_train = y.loc[train_idx]
    X_test = X.drop(train_idx)
    y_test = y.drop(train_idx)

    return X_train,X_test,y_train,y_test

X_train,X_test,y_train,y_test = train_test_split(X,y)

def scale(x):
    return (x - x.mean())/x.std()

def primal_objective(X,y,w,alpha):
    RSS = sum((np.dot(X,w)-y)**2)
    penalty = alpha * sum(np.abs(w))

    n = X.shape[0]

    return (1/(2*n))*RSS + penalty

def current_residual(X,y,w):
    return y - np.dot(X,w)

def dual_variable(X,r,alpha):
    n = X.shape[0]

    max = 0
    for col in X.columns:
        correlation = np.abs(X[col].dot(r))

        if correlation > max:
            max = correlation

    if max <= n*alpha :
        c = 1
    else :
        c = (n * alpha)/max

    return c * r

def dual_objective(y,v):
    n = y.shape[0]

    return (1/2*n)*(np.dot(y.T,v) - sum(v**2))

def duality_gap(pw,dv):
    return pw - dv

def fit(X,y,tol=1e-4,alpha=1,max_iter=1000):
    X_scaled = scale(X)

    W = np.zeros(X.shape[1])

    for i in range(max_iter):
        penalty = alpha * sum(np.abs(W))
        for col, i in zip(X.columns, range(len(W))) :
            x = X.drop(columns=[col])
            w_subset = np.delete(W,i)

            r = y - np.dot(x,w_subset)
            z = np.dot(X[col].T,r)


            STO = np.sign(z) * max(0,np.abs(z)-penalty)

            w_new = STO / sum(X[col]**2)

            W[i] = w_new

        pw = primal_objective(X,y,W,alpha)
        r = current_residual(X,y,W)
        v = dual_variable(X,r,alpha)
        dv = dual_objective(y,v)
        G = duality_gap(pw,dv)

        if G < tol * (sum(y**2))/X.shape[0] :
            break
        else :
            continue

    b = y.mean() - np.dot(W,X.mean())

    return W,b

W,b = fit(X_train,y_train,alpha=160)

def predict(X):
    return np.dot(X,W) + b

y_predict = predict(X_test)

RSS = sum((y_test - y_predict)**2)
RSE = np.sqrt(RSS/(len(X_test) - 2))
TSS = sum((y_test - y_test.mean())**2)
R2 = 1 - (RSS/TSS)

print(f"RSS : {RSS}")
print(f"RSE : {RSE}")
print(f"TSS : {TSS}")
print(f"R2 : {R2}")
```
