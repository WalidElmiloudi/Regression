## 2. What Is Elastic Net Regression?

Elastic Net is a regularized version of Multiple Linear Regression that combines two regularization techniques:

- **L1 Regularization (Lasso):** Encourages sparsity and can force some coefficients to become exactly zero.
- **L2 Regularization (Ridge):** Shrinks coefficients toward zero without generally making them exactly zero.

By combining both penalties, Elastic Net benefits from the strengths of Ridge and Lasso.

The mathematical representation of the prediction model remains:

\[
$\hat{y} = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b$
\]

However, the optimization process includes both L1 and L2 penalties.

**Why do we need Elastic Net Regression?**

- Overfitting: Helps control model complexity and improve generalization.
- Multicollinearity: Handles situations where independent variables are highly correlated.
- Feature selection: Can eliminate less useful features by setting their coefficients to zero.
- Coefficient stability: Combines L1 sparsity with L2 coefficient shrinkage.

## 3. Mathematical Foundations

### 3.1. Mean Squared Error (MSE)

Ordinary Linear Regression minimizes the Mean Squared Error:

\[
$MSE(W,b) = \frac{1}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})^2$
\]

Where:

- \$(m\$): Number of training examples.
- \$(\hat{y}^{(i)}\$): Predicted value.
- \$(y^{(i)}\$): Actual target value.
- \$(W\$): Vector of model weights.
- \$(b\$): Bias (intercept).

### 3.2. L1 Regularization (Lasso)

L1 Regularization penalizes the sum of the absolute values of the model coefficients:

\[
$L1(W)=\sum_{j=1}^{n}|w_j|$
\]

Its main characteristic is that it can force some coefficients to become exactly zero, allowing the model to perform automatic feature selection.

### 3.3. L2 Regularization (Ridge)

L2 Regularization penalizes the sum of the squared model coefficients:

\[
$L2(W)=\sum_{j=1}^{n}w_j^2$
\]

Its main characteristic is that it shrinks coefficients toward zero, helping reduce excessively large weights and improve model stability.

### 3.4. Elastic Net Cost Function

Elastic Net combines both penalties into a single objective function:

\[
$J(W,b)=$
$\frac{1}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})^2$
+
$\lambda\rho\sum_{j=1}^{n}|w_j|$
+
$\frac{\lambda(1-\rho)}{2}\sum_{j=1}^{n}w_j^2$
\]

Where:

- \$(MSE\$): Measures prediction error.
- \$(\lambda\$): Controls the overall regularization strength.
- \$(\rho\$): Controls the balance between L1 and L2 regularization.
- \$(W\$): Vector of model weights.
- \$(b\$): Bias, which is not penalized.

**Understanding the parameters:**

- When \$(\lambda=0\$), Elastic Net becomes ordinary linear regression.
- When \$(\rho=1\$), Elastic Net becomes Lasso Regression.
- When \$(\rho=0\$), Elastic Net becomes Ridge Regression.
- When \$(0<\rho<1\$), both L1 and L2 penalties are applied.

The regularization parameter controls how strongly the model penalizes large coefficients, while the mixing parameter controls how much each regularization technique contributes.

### 3.5. Elastic Net Using Gradient Descent

To optimize Elastic Net, we need to calculate the gradients of its cost function.

First, calculate predictions:

\[
$\hat{y}=XW+b$
\]

The cost function is:

\[
$J(W,b)=$
$\frac{1}{m}\|XW+b-y\|_2^2$
+
$\lambda\rho\|W\|_1$
+
$\frac{\lambda(1-\rho)}{2}\|W\|_2^2$
\]

**Weight subgradient:**

\[
$\frac{\partial J}{\partial W}$
=
$\frac{2}{m}X^T(\hat{y}-y)$
+
$\lambda\rho\operatorname{sign}(W)$
+
$\lambda(1-\rho)W$
\]

**Bias gradient:**

\[
$\frac{\partial J}{\partial b}$
=
$\frac{2}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})$
\]

**Parameter updates:**

\[
$W := W-\alpha\frac{\partial J}{\partial W}$
\]

\[
$b := b-\alpha\frac{\partial J}{\partial b}$
\]

Where \$(\alpha\$) represents the learning rate.

**Important:** The L1 component is not differentiable at zero. The sign function provides a valid subgradient, with zero being one possible choice at the origin. This basic subgradient approach may not produce exact zero coefficients in a finite number of iterations.

For more effective optimization and exact sparsity, we can also explore proximal Gradient Descent using the soft-thresholding operator.

```python
import pandas as pd
import numpy as np
import math
from pathlib import Path

cwd = Path.cwd()
root = cwd.parent
data_dir = root / "Datasets" / "Elastic-net-regression"

df = pd.read_csv(data_dir / "elasticnet_regression_dataset.csv")

print(df.info())
print(df.describe())
print(df.shape)
```

```python
y = df["performance_score"]
X = df.drop(columns=["performance_score"])

def train_test_split(X,y,test_size=0.2):
    train_idx = X.sample(frac=1-test_size,random_state=42).index

    X_train = X.loc[train_idx]
    y_train = y.loc[train_idx]
    X_test = X.drop(train_idx)
    y_test = y.drop(train_idx)

    return X_train,X_test,y_train,y_test

X_train,X_test,y_train,y_test = train_test_split(X,y)

def scale(x):
    return (x - x.mean())/x.std()

def primal_objective(X,y,w,alpha,l1_ratio=0):
    RSS = sum((np.dot(X,w)-y)**2)
    penalty = (alpha * l1_ratio * sum(np.abs(w))) + ((alpha * (1 - l1_ratio))/2) * sum(w**2)

    n = X.shape[0]

    return (1/(2*n))*RSS + penalty

def current_residual(X,y,w):
    return y - np.dot(X,w)

def dual_variable(X,r,alpha):
    n = X.shape[0]

    max = 0
    for col in X.columns:
        correlation = np.abs(X[col].dot(r))

        if correlation > max:
            max = correlation

    if max <= n*alpha :
        c = 1
    else :
        c = (n * alpha)/max

    return c * r

def dual_objective(y,v):
    n = y.shape[0]

    return (1/2*n)*(np.dot(y.T,v) - sum(v**2))

def duality_gap(pw,dv):
    return pw - dv

def fit(X,y,tol=1e-4,alpha=1,max_iter=1000,l1_ratio=0):
    X_scaled = scale(X)

    W = np.zeros(X.shape[1])

    for i in range(max_iter):
        for col, i in zip(X.columns, range(len(W))) :
            x = X.drop(columns=[col])
            w_subset = np.delete(W,i)

            r = y - np.dot(x,w_subset)
            z = np.dot(X[col].T,r)


            STO = np.sign(z) * max(0,np.abs(z)-(alpha*l1_ratio))

            w_new = STO / (sum(X[col]**2)+(alpha * (1 - l1_ratio)))

            W[i] = w_new

        pw = primal_objective(X,y,W,alpha)
        r = current_residual(X,y,W)
        v = dual_variable(X,r,alpha)
        dv = dual_objective(y,v)
        G = duality_gap(pw,dv)

        if G < tol * (sum(y**2))/X.shape[0] :
            break
        else :
            continue

    b = y.mean() - np.dot(W,X.mean())

    return W,b

W,b = fit(X_train,y_train,alpha=0,l1_ratio=0)

def predict(X):
    return np.dot(X,W) + b

y_predict = predict(X_test)

RSS = sum((y_test - y_predict)**2)
RSE = np.sqrt(RSS/(len(X_test) - 2))
TSS = sum((y_test - y_test.mean())**2)
R2 = 1 - (RSS/TSS)

print(f"RSS : {RSS}")
print(f"RSE : {RSE}")
print(f"TSS : {TSS}")
print(f"R2 : {R2}")
```
