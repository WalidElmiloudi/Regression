## 2. What Is Ridge Regression?

Ridge Regression is a regularized version of Multiple Linear Regression.

Like ordinary linear regression, it models the relationship between multiple independent variables (features) and one dependent variable (target).

However, it introduces an additional penalty term into the cost function to discourage excessively large coefficients.

The mathematical representation remains:

\[
$\hat{y} = w_1x_1 + w_2x_2 + \cdots + w_nx_n + b$
\]

The main difference is that Ridge Regression penalizes large weights during model optimization.

**Why do we need Ridge Regression?**

Ordinary Linear Regression can suffer from:

- Overfitting: The model learns the training data too closely and performs poorly on unseen data.
- Multicollinearity: Highly correlated features can produce unstable and excessively large coefficients.
- High model variance: Small changes in training data can cause significant changes in the estimated coefficients.

Ridge Regression addresses these problems by introducing L2 Regularization.

## 3. Mathematical Foundations

### 3.1. Mean Squared Error (MSE)

Ordinary Linear Regression minimizes the Mean Squared Error:

\[
$MSE(W,b) = \frac{1}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})^2$
\]

Where:

- \$(m\$): Number of training examples.
- \$(\hat{y}^{(i)}\$): Predicted value.
- \$(y^{(i)}\$): Actual value.
- \$(W\$): Vector of model weights.
- \$(b\$): Bias.

### 3.2. L2 Regularization Penalty

Ridge Regression introduces a penalty based on the squared values of the model weights:

\[
$L2(W) = \sum_{j=1}^{n}w_j^2$
\]

The regularization penalty discourages the model from assigning excessively large weights to its features.

### 3.3. Ridge Regression Cost Function

The Ridge Regression objective combines the prediction error and the regularization penalty:

\[
$J(W,b) = MSE(W,b) + \lambda\sum_{j=1}^{n}w_j^2$
\]

Where:

- \$(MSE\$): Measures prediction error.
- \$(\lambda\$): Regularization parameter (controls the strength of the penalty).
- \$(W\$): Model weights.

**Important:**

- When \$(\lambda = 0\$), Ridge Regression becomes ordinary linear regression.
- When \$(\lambda\$) increases, the model penalizes large coefficients more strongly.
- Ridge Regression shrinks coefficients toward zero but generally does not make them exactly zero.
- The bias (intercept) is not penalized.

### 3.4. Ridge Regression Using the Analytical Solution

For the analytical approach, we can calculate the optimal coefficients using a modified OLS equation:

\[
$W = (X^TX+\lambda I)^{-1}X^Ty$
\]

Where:

- \$(X\$): Feature matrix.
- \$(X^T\$): Transpose of the feature matrix.
- \$(I\$): Identity matrix.
- \$(\lambda\$): Regularization strength.
- \$(y\$): Target vector.

The bias should be handled separately or through an augmented matrix with zero regularization applied to its intercept coefficient.

This modification makes the matrix more stable when features are highly correlated.

### 3.5. Ridge Regression Using Gradient Descent

We can also optimize Ridge Regression using Gradient Descent.

First, we calculate predictions:

\[
$\hat{y}=XW+b$
\]

Then, we calculate the cost function:

\[
$J(W,b)=\frac{1}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})^2+\lambda\sum_{j=1}^{n}w_j^2$
\]

**Weight gradient:**

\[
$\frac{\partial J}{\partial W} =$
$\frac{2}{m}X^T(\hat{y}-y)+2\lambda W$
\]

**Bias gradient:**

\[
$\frac{\partial J}{\partial b} =$
$\frac{2}{m}\sum_{i=1}^{m}(\hat{y}^{(i)}-y^{(i)})$
\]

**Parameter updates:**

\[
$W := W-\alpha\frac{\partial J}{\partial W}$
\]

\[
$b := b-\alpha\frac{\partial J}{\partial b}$
\]

Where \$(\alpha\$) represents the learning rate.

Notice that the weight gradient contains an additional term, \$(2\lambda W\$), which pushes the weights toward zero.

The bias gradient does not contain this penalty because the intercept is not regularized.

```python
import pandas as pd
import numpy as np
import math
from pathlib import Path

cwd = Path.cwd()
root = cwd.parent
data_dir = root / "Datasets" / "Ridge-regression"

df = pd.read_csv(data_dir / "mydata.csv")

print(df.info())
print(df.describe())
print(df.shape)
```

```python
# df = df.drop(columns=["Unnamed: 0"])

y = df["y"]
X = df.drop(columns=["y"])

def train_test_split(X,y,test_size=0.2):
    train_idx = X.sample(frac=1-test_size, random_state=42).index

    X_train = X.loc[train_idx]
    y_train = y.loc[train_idx]
    X_test = X.drop(train_idx)
    y_test = y.drop(train_idx)

    return X_train,X_test,y_train,y_test

X_train,X_test,y_train,y_test = train_test_split(X,y)

def scale(x):
    return (x-x.mean())/x.std()

def fit(X,y,C=0):
    X_scaled = scale(X)

    inv_term = np.linalg.inv(np.dot(X_scaled.T, X_scaled) + C * np.eye(np.dot(X_scaled.T,X_scaled).shape[0]))
    W = np.dot(inv_term,np.dot(X_scaled.T,y))

    b = y.mean() - np.dot(W,X_scaled.mean())

    return W , b

W , b = fit(X_train,y_train,C=1)

def predict(X):
    X_scaled = scale(X)
    return np.dot(X_scaled,W) + b

y_predict = predict(X_test)

RSS = sum((y_test - y_predict)**2)
RSE = np.sqrt(RSS/(len(X_test) - 2))
TSS = sum((y_test - y_test.mean())**2)
R2 = 1 - (RSS/TSS)

print(f"RSS : {RSS}")
print(f"RSE : {RSE}")
print(f"TSS : {TSS}")
print(f"R2 : {R2}")
```
